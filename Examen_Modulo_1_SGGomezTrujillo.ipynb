#Creación de Clase Tienda On Line

```python
class TiendaOnLine:
    
    def __init__(self, nombre):
        self.nombre = nombre
        self.inventario = []                                # Lista de diccionarios: cada dict es {"nombre": str, "precio": float, "cantidad": int}
        self.ventas_totales = 0
    
    def agregar_producto(self, producto, precio, cantidad):
        # Validaciones
        if not producto or not precio or not cantidad:
            print("Debes ingresar nombre, precio y cantidad del producto.\nEjecuta nuevamente.")
            return
        if precio <= 0:                                     #posibilidad de mejora, imposibilidad de ingreso de string.
            print("Debes ingresar un precio mayor a 0.")
            return
        if cantidad <= 0 or not isinstance(cantidad,int):   #posibilidad de mejora, imposibilidad de ingreso de string.
            print("Debes ingresar un número entero, mayor a 0.")
            return

        # Busqueda en el inventario
        for item in self.inventario:
            if item["nombre"] == producto.title():
                item["cantidad"] += cantidad
                print(f"Se actualizaron las unidades del producto '{producto}'. Hay {item['cantidad']:,.0f} unidad(es) disponible(s).")
                return
        
        # Si no existe, agregar nuevo producto
        nuevo_producto = {"nombre": producto.title(), "precio": precio, "cantidad": cantidad}
        self.inventario.append(nuevo_producto)
        print(f"El producto '{producto} ' fue agregado al inventario.")  #sin el title(), replica el nombre ingresado para corroborar con el usuario que fuera error de tipeo.
        print(self.inventario)
    
    def ver_inventario(self):
        if not self.inventario:
            print("El inventario está vacío.")
            #return
        
        for producto in self.inventario:
            print(f"Nombre: {producto}, Precio: ${producto['precio']:,.2f}, Cantidad: {producto['cantidad']:,.0f}")
        
```

```python
tester=TiendaOnLine("tester")
```

```python
tester.ver_inventario()
```

```python
tester.agregar_producto('jean',25.5,420)
```

```python
tester.ver_inventario()
```

```python
tester.agregar_producto('remera',7.5,450)
tester.agregar_producto('vestido',12.48,14)
tester.agregar_producto('campera',34,22)
```

```python
tester.ver_inventario()
```
